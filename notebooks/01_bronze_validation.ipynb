# Step 1 - Read Bronze Table

Read the Bronze events table from Unity Catalog.

```python
bronze_df = spark.table("workspace.bronze.events_raw")
```

# Step 2 - Inspect Schema

Inspect the schema to verify column names and data types.

```python
bronze_df.printSchema()
```

# Step 3 - Preview Data

Preview the raw Bronze data.

```python
display(bronze_df)
```

# Step 4 - Count Records

Verify the total number of records.

```python
bronze_df.count()
```

# Step 5 - Check Missing Values

Inspect column-level missing values.

**Note**

`checkout_id` is expected to be NULL for non-checkout events by design.

This validation checks column completeness only.

It does **not** validate the "Missing Events" data quality scenario, where entire rows are intentionally removed during synthetic data generation.

```python
from pyspark.sql.functions import col, count, when

display(
    bronze_df.select([
        count(
            when(col(c).isNull(), c)
        ).alias(c)
        for c in bronze_df.columns
    ])
)
```

# Step 6 - Event Distribution

Inspect event type distribution.

The result should be consistent with the synthetic data generator output.

```python
display(
    bronze_df
        .groupBy("event_name")
        .count()
        .orderBy("event_name")
)
```

# Step 7 - Country Distribution

Inspect country distribution.

```python
display(
    bronze_df
        .groupBy("country")
        .count()
        .orderBy("country")
)
```

# Step 8 - Platform Distribution

Inspect platform distribution.

```python
display(
    bronze_df
        .groupBy("platform")
        .count()
        .orderBy("platform")
)
```
